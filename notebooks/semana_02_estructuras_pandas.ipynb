## Actividad 3.2 - Listas y diccionarios

En esta actividad se utilizan estructuras básicas de Python para almacenar,
consultar y procesar información de productos.

Se desarrollan operaciones de recorrido, filtrado y cálculo de valores.

```python
# Actividad 3.2 - Listas y diccionarios

# Lista de productos
productos = ["Cemento", "Arena", "Triturado", "Varilla", "Ladrillo"]

# Mostrar todos los productos
print("LISTA DE PRODUCTOS")

for producto in productos:
    print("-", producto)

# Consultar un producto específico
print("\nPrimer producto:", productos[0])

# Cantidad de productos
print("Total de productos:", len(productos))
```

```python
# Diccionario con información de un producto

producto = {
    "nombre": "Cemento",
    "precio": 35000,
    "cantidad": 20,
    "categoria": "Materiales de construcción"
}

print("INFORMACIÓN DEL PRODUCTO")

print("Nombre:", producto["nombre"])
print("Precio:", producto["precio"])
print("Cantidad:", producto["cantidad"])
print("Categoría:", producto["categoria"])

# Calcular el valor total del inventario
valor_total = producto["precio"] * producto["cantidad"]

print("Valor total:", valor_total)
```

### Filtrado y procesamiento de información

En este ejercicio se utiliza una lista de diccionarios para representar
un inventario de materiales de construcción.

Se aplican filtros, recorridos y operaciones matemáticas para obtener
información útil a partir de los datos almacenados.

```python
# Inventario representado mediante una lista de diccionarios

inventario = [
    {"nombre": "Cemento", "precio": 35000, "cantidad": 20},
    {"nombre": "Arena", "precio": 80000, "cantidad": 5},
    {"nombre": "Triturado", "precio": 90000, "cantidad": 8},
    {"nombre": "Varilla", "precio": 25000, "cantidad": 15},
    {"nombre": "Ladrillo", "precio": 1200, "cantidad": 500}
]

# Mostrar el inventario completo
print("INVENTARIO DE MATERIALES")

for material in inventario:
    print(
        material["nombre"],
        "- Precio:", material["precio"],
        "- Cantidad:", material["cantidad"]
    )

# Filtrar materiales cuyo precio sea mayor a 30000
print("\nMATERIALES CON PRECIO MAYOR A $30.000")

for material in inventario:
    if material["precio"] > 30000:
        print(material["nombre"], "-", material["precio"])

# Calcular el valor total del inventario
total_inventario = 0

for material in inventario:
    subtotal = material["precio"] * material["cantidad"]
    total_inventario += subtotal

print("\nVALOR TOTAL DEL INVENTARIO:", total_inventario)
```

## Actividad 3.3 - Operaciones con NumPy

En esta actividad se utiliza la biblioteca NumPy para realizar
operaciones matemáticas y estadísticas sobre los precios de
materiales de construcción.

Se calculan la suma, el promedio, el valor máximo, el valor
mínimo y la desviación estándar.

```python
# Actividad 3.3 - Operaciones con NumPy

import numpy as np

# Precios de los materiales
precios = np.array([35000, 80000, 90000, 25000, 1200])

print("PRECIOS DE LOS MATERIALES")
print(precios)

# Operaciones estadísticas
print("\nRESULTADOS ESTADÍSTICOS")

print("Cantidad de precios:", len(precios))
print("Suma de precios:", np.sum(precios))
print("Promedio:", np.mean(precios))
print("Precio máximo:", np.max(precios))
print("Precio mínimo:", np.min(precios))
print("Desviación estándar:", np.std(precios))
```

```python
# Filtrar precios mayores a 30000

precios_altos = precios[precios > 30000]

print("PRECIOS MAYORES A $30.000")
print(precios_altos)

# Aplicar un incremento del 10 %
precios_incrementados = precios * 1.10

print("\nPRECIOS CON INCREMENTO DEL 10 %")
print(precios_incrementados)
```

## Actividad 3.4 - Series y DataFrames con Pandas

En esta actividad se utiliza la biblioteca Pandas para organizar
información mediante estructuras Series y DataFrames.

Se construye una tabla de inventario de materiales de construcción,
se consultan sus registros y se realizan operaciones básicas de análisis.

```python
# Actividad 3.4 - Series y DataFrames con Pandas

import pandas as pd

# Crear una Serie con los precios de los materiales
serie_precios = pd.Series(
    [35000, 80000, 90000, 25000, 1200],
    index=["Cemento", "Arena", "Triturado", "Varilla", "Ladrillo"]
)

print("SERIE DE PRECIOS")
print(serie_precios)

# Calcular estadísticas
print("\nPROMEDIO:", serie_precios.mean())
print("PRECIO MÁXIMO:", serie_precios.max())
print("PRECIO MÍNIMO:", serie_precios.min())
```

```python
# Crear un DataFrame a partir del inventario anterior

df_inventario = pd.DataFrame(inventario)

# Calcular el valor total por material
df_inventario["valor_total"] = (
    df_inventario["precio"] * df_inventario["cantidad"]
)

print("INVENTARIO DE MATERIALES")

display(df_inventario)
```

```python
# Consultar los primeros registros
print("PRIMEROS REGISTROS")
display(df_inventario.head())

# Consultar información general
print("\nINFORMACIÓN DEL DATAFRAME")
df_inventario.info()

# Obtener estadísticas descriptivas
print("\nESTADÍSTICAS DESCRIPTIVAS")
display(df_inventario.describe())

# Filtrar materiales con precio mayor a 30000
print("\nMATERIALES CON PRECIO MAYOR A $30.000")

materiales_costosos = df_inventario[
    df_inventario["precio"] > 30000
]

display(materiales_costosos)
```

## Actividad 3.5 - Lectura y escritura de archivos CSV y Excel

de un DataFrame a archivos CSV y Excel.

verificar que la información se haya almacenado correctamente.

Los archivos se organizan en las carpetas data/raw y
data/processed del proyecto.

## Actividad 3.6 - Análisis de datos con Pandas

En esta actividad se realiza un análisis del inventario de materiales
de construcción utilizando Pandas.

Se calculan indicadores como el valor total del inventario,
el precio promedio, el material más costoso y el material
con mayor cantidad disponible.

El objetivo es transformar los datos almacenados en información
útil para la toma de decisiones.

```python
# Actividad 3.6 - Análisis de datos

import pandas as pd

# Trabajar con los datos importados
df_analisis = df_csv.copy()

# Calcular indicadores generales
total_materiales = len(df_analisis)
valor_inventario = df_analisis["valor_total"].sum()
precio_promedio = df_analisis["precio"].mean()

# Identificar materiales destacados
material_costoso = df_analisis.loc[
    df_analisis["precio"].idxmax()
]

material_mayor_cantidad = df_analisis.loc[
    df_analisis["cantidad"].idxmax()
]

# Mostrar resultados
print("ANÁLISIS GENERAL DEL INVENTARIO")

print("Total de tipos de materiales:", total_materiales)
print("Valor total del inventario: $", valor_inventario)
print("Precio promedio por unidad: $", precio_promedio)

print("\nMATERIAL MÁS COSTOSO")
print("Nombre:", material_costoso["nombre"])
print("Precio: $", material_costoso["precio"])

print("\nMATERIAL CON MAYOR CANTIDAD")
print("Nombre:", material_mayor_cantidad["nombre"])
print("Cantidad:", material_mayor_cantidad["cantidad"])
```
